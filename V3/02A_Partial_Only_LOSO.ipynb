# NB02A — Partial-only ResNet18 U-Net: four LOSO models and independent 30-circle validation

Train one **ImageNet-pretrained ResNet18 U-Net** per held-out site using only *non-exhaustive* tree labels from the other three sites. First report held-out annotated-tree pixel ROC-AUC/AP; then evaluate the **30 withheld-site exhaustive circles** for pixel discrimination and Musanga fractional canopy cover. Exhaustive circle boundaries are exclusion buffers throughout training; their labels are evaluation-only. MOS 3–4 is excluded. Existing compatible Baego checkpoint is reused.

## 02 — Load annotations and ImageNet-pretrained ResNet18 U-Net

```python
def pieces(g):
    if g.geom_type in ('Polygon','MultiPolygon'):return [g]
    if hasattr(g,'geoms'):return [p for q in g.geoms for p in pieces(q)]
    return []
def polygons(path):
    src=gpd.read_file(path);inds=[];shapes=[]
    for i,g in enumerate(src.geometry):
        if g is None or g.is_empty:continue
        if not g.is_valid:
            try:
                fixed=unary_union(pieces(make_valid(g)))
                change=abs(fixed.area-g.area)/max(fixed.area,g.area,1e-12)
                if change>(.01 if Path(path) in (Path(A['exhaustive_annotations']['complete_roi_boundaries']),Path(A['exhaustive_annotations']['positive_polygons']),Path(A['roi'])) else .05) or not fixed.is_valid:continue
                g=fixed
            except Exception:continue
        inds.append(i);shapes.append(g)
    x=src.iloc[inds].copy().reset_index(drop=True)
    x.geometry=shapes
    x['original_row']=inds
    return x
roi=polygons(A['roi'])
circles=polygons(A['exhaustive_annotations']['complete_roi_boundaries'])
positive=polygons(A['exhaustive_annotations']['positive_polygons'])
assert len(circles)==34 and len(positive)==1406, (
    f'Unexpected reference geometries: circles={len(circles)} (expected 34), '
    f'Musanga polygons={len(positive)} (expected 1406); do not train until reconciled')
circle_ids=set(links.circle_id.astype(int))
circles=circles[circles.original_row.isin(circle_ids)].copy()
tree={}
for site,path in A['shared_annotations'].items():
    t=polygons(path)
    cls=pd.to_numeric(t.Class_ID,errors='coerce')
    t['class01']=np.where(cls.eq(1),1,np.where(cls.notna(),0,-1))
    tree[site]=t
geo={}
for name,m in mosaics.items():
    with rasterio.open(m['path']) as src:
        c=circles.to_crs(src.crs).copy()
        c['fold']=c.original_row.map(links.set_index('circle_id').fold)
        c['owner']=c.original_row.map(links.set_index('circle_id').mosaic)
        geo[name]=dict(site=m['site'],path=m['path'],transform=src.transform,
            crs=src.crs,roi=roi.to_crs(src.crs),
            circles=c,positive=positive.to_crs(src.crs),trees=tree[m['site']].to_crs(src.crs))

```

```python
class ResNet18UNet(nn.Module):
    """Original main-branch U-Net: ImageNet-pretrained ResNet18 encoder, RGB in, 1 logit out."""
    def __init__(self):
        super().__init__()
        import segmentation_models_pytorch as smp
        self.net=smp.Unet(encoder_name='resnet18',encoder_weights='imagenet',
                          in_channels=3,classes=1)
    def forward(self,x):
        return self.net(x)[:,0]

class CropData(Dataset):
    def __init__(self,rows,augment=False):self.rows=rows;self.augment=augment
    def __len__(self):return len(self.rows)
    def __getitem__(self,i):
        d=self.rows[i];image=d['rgb'];y=d['y'];w=d['w']
        if self.augment:
            rng=np.random.default_rng(SEED+i+int(torch.randint(0,1000000,(1,)).item()))
            k=int(rng.integers(4));image=np.rot90(image,k);y=np.rot90(y,k);w=np.rot90(w,k)
            if rng.random()<.5:image=image[:,::-1];y=y[:,::-1];w=w[:,::-1]
        x=torch.tensor(np.ascontiguousarray(image.transpose(2,0,1)),dtype=torch.float32)
        return x,torch.tensor(np.ascontiguousarray(y)),torch.tensor(np.ascontiguousarray(w))
def masked_loss(logits,y,w):
    loss=F.binary_cross_entropy_with_logits(logits,y,reduction='none')
    return (loss*w).sum()/w.sum().clamp_min(1.)
print('ResNet18 U-Net ready (native 256×256 RGB crops).')


```

## 03 — Strict training exclusion: all 34 exhaustive circles (including MOS cross-mosaic buffers)

```python
HOLDOUT=set(links.circle_id.astype(int))
DEV=set()
split_df=links[['circle_id','site','mosaic','cover_pct']].copy()
split_df['split']='holdout'
BLOCKED={}
for name,d in geo.items():
    this=circles.to_crs(d['crs'])
    # 15m exclusion across the complete site, not only the mosaic assigned to a circle.
    site_ids=set(links.loc[links.site==d['site'],'circle_id'].astype(int))
    BLOCKED[name]=[g.buffer(15.) for g in this[this.original_row.isin(site_ids)].geometry]
assert len(HOLDOUT)==30
print('All',len(HOLDOUT),'circles excluded from training, across MOS mosaic boundaries.')
```

## 04 — Cached RGB windows (no full GeoTIFF copies)

Directly read only requested windows from Drive. Persist fingerprinted chip caches and reuse compatible fold checkpoints. This changes I/O only, not the training data or model. Progress is printed by mosaic, chunk and epoch.


```python
from rasterio.windows import Window
from torch.utils.data import Dataset,DataLoader
candidate=pd.read_csv(REF/'01_candidate_windows.csv')
candidate=candidate[candidate.mosaic.isin(mosaics)].copy()
all_circle_buffers={}
for name,d in geo.items():
    cs=circles.to_crs(d['crs'])
    all_circle_buffers[name]=[g.buffer(15.) for g in cs.geometry]
    d['dev_circles']=cs[cs.original_row.isin(DEV)]
    d['holdout_circles']=cs[cs.original_row.isin(HOLDOUT)]
    d['positive_local']=positive.to_crs(d['crs'])
    d['trees_local']=tree[d['site']].to_crs(d['crs'])
    d['tree_positive']=d['trees_local'].loc[d['trees_local'].class01==1]
    d['tree_negative']=d['trees_local'].loc[d['trees_local'].class01==0]
    d['roi_local']=roi.to_crs(d['crs'])

def burn_window(geometries,name,win):
    """Rasterize geometries intersecting a crop, using the spatial index.

    Accept GeoSeries OR GeoDataFrame; always iterate actual Shapely geometries,
    never GeoDataFrame column names.
    """
    d=geo[name]
    series=geometries.geometry if isinstance(geometries,gpd.GeoDataFrame) else geometries
    assert isinstance(series,gpd.GeoSeries),type(series)
    bb=box(*rasterio.windows.bounds(win,d['transform']))
    ids=series.sindex.query(bb,predicate='intersects')
    shape=(int(win.height),int(win.width))
    if len(ids)==0:return np.zeros(shape,dtype=bool)
    selected=series.iloc[ids]
    return rasterize(((g,1) for g in selected.values
                      if g is not None and not g.is_empty),
        out_shape=shape,transform=rasterio.windows.transform(win,d['transform']),
        fill=0,dtype='uint8').astype(bool)

_ACTIVE_RASTER=None
_ACTIVE_NAME=None

def rgb_for_window(name,win):
    if _ACTIVE_RASTER is not None and _ACTIVE_NAME==name:
        raw=_ACTIVE_RASTER.read([1,2,3],window=win,masked=True)
    else:
        with rasterio.open(geo[name]['path']) as ds:
            raw=ds.read([1,2,3],window=win,masked=True)
    valid=~np.any(np.ma.getmaskarray(raw),axis=0)
    a=raw.filled(0).astype('float32')
    valid &= np.any(a!=0,axis=0)&np.any(a!=255,axis=0)
    valid &= ~((a[0]==0)&(a[1]==0)&(a[2]==255))
    return np.clip(a/255.,0,1).transpose(1,2,0).astype('float32'),valid

def training_crop(rec,model):
    name=rec.mosaic;d=geo[name]
    if rec.height<CROP or rec.width<CROP:return None
    rr=int(rec.row)+(int(rec.height)-CROP)//2
    cc=int(rec.col)+(int(rec.width)-CROP)//2
    win=Window(cc,rr,CROP,CROP)
    bb=box(*rasterio.windows.bounds(win,d['transform']))
    if any(bb.intersects(g) for g in BLOCKED[name]):return None
    if model=='M1' and any(bb.intersects(g) for g in all_circle_buffers[name]):return None
    if model=='M2' and rec.kind=='exhaustive' and int(rec.circle_id) not in DEV:return None
    inside=burn_window(d['roi_local'].geometry,name,win)
    pos_tree=burn_window(d['tree_positive'],name,win)
    neg_tree=burn_window(d['tree_negative'],name,win)
    if model=='M1':
        full=np.zeros_like(inside);pos_ex=np.zeros_like(inside)
    else:
        full=burn_window(d['dev_circles'].geometry,name,win)&inside
        pos_ex=burn_window(d['positive_local'].geometry,name,win)&full
    conflict=(pos_tree&neg_tree)| (pos_ex&neg_tree) | (full&~pos_ex&pos_tree)
    target=(pos_ex|(~full&pos_tree)).astype('float32')
    weight=(full|((pos_tree|neg_tree)&~full))&inside&~conflict
    img,valid=rgb_for_window(name,win)
    weight &= valid
    if int(weight.sum())<250:return None
    return dict(image=img,target=target,weight=weight.astype('float32'),
                site=d['site'],kind=rec.kind,mosaic=name)

def _pack_chips(path,items,signature):
    """Atomic, compressed uint8 cache of already-prepared chips."""
    if not items:return
    tmp=path.with_name(path.name+'.tmp.npz')
    try:
        np.savez_compressed(tmp,key=np.array(signature),
            rgb=np.stack([np.rint(np.clip(r['image'],0,1)*255).astype('uint8') for r in items]),
            target=np.stack([r['target'].astype('uint8') for r in items]),
            weight=np.stack([r['weight'].astype('uint8') for r in items]),
            site=np.array([r['site'] for r in items],dtype='U12'),
            kind=np.array([r['kind'] for r in items],dtype='U16'))
        tmp.replace(path)
    finally:
        tmp.unlink(missing_ok=True)

def _unpack_chips(path,signature):
    with np.load(path,allow_pickle=False) as z:
        assert str(z['key'].item())==signature
        rgb=z['rgb'];target=z['target'];weight=z['weight']
        sites=z['site'].astype(str);kinds=z['kind'].astype(str)
        assert rgb.ndim==4 and rgb.shape[1:]==(CROP,CROP,3)
        assert target.shape==weight.shape==rgb.shape[:3]
        return [dict(image=rgb[i].astype('float32')/255.,
                     target=target[i].astype('float32'),
                     weight=weight[i].astype('float32'),
                     site=str(sites[i]),kind=str(kinds[i]))
                for i in range(len(rgb))]

import shutil
from contextlib import contextmanager
LOCAL_STAGE=Path('/content/musanga_02f_stage')
LOCAL_STAGE.mkdir(parents=True,exist_ok=True)
MIN_FREE_AFTER_STAGE=2*1024**3
USE_LOCAL_STAGE=False

@contextmanager
def staged_mosaic(name):
    """Stage one raster on SSD if it fits; use one open raster handle per mosaic."""
    global _ACTIVE_RASTER,_ACTIVE_NAME
    source=Path(geo[name]['path'])
    assert source.exists(),source
    free=shutil.disk_usage('/content').free
    needed=source.stat().st_size
    local=LOCAL_STAGE/(name.replace('/','_'))
    can_stage=USE_LOCAL_STAGE and free>=needed+MIN_FREE_AFTER_STAGE
    selected_path=source
    if can_stage:
        print(f'STAGE {name}: {needed/1e9:.2f} GB from Drive to local SSD '
              f'(available {free/1e9:.1f} GB)',flush=True)
        t0=time.perf_counter()
        tmp=local.with_name(local.name+'.part')
        try:
            with source.open('rb') as inp,tmp.open('wb') as out:
                shutil.copyfileobj(inp,out,length=16*1024*1024)
            assert tmp.stat().st_size==needed
            tmp.replace(local)
            selected_path=local
            print(f'STAGED {name} in {time.perf_counter()-t0:.1f}s',flush=True)
        except Exception as exc:
            tmp.unlink(missing_ok=True)
            local.unlink(missing_ok=True)
            print(f'Local copy failed; using original Drive image: {exc}',flush=True)
    else:
        print(f'STAGE SKIPPED {name}: image {needed/1e9:.2f} GB, '
              f'free SSD {free/1e9:.2f} GB; using one persistent Drive handle',
              flush=True)
    assert _ACTIVE_RASTER is None,'Another mosaic is already open.'
    try:
        with rasterio.open(selected_path) as ds:
            assert ds.count>=3 and ds.transform==geo[name]['transform']
            assert ds.crs==geo[name]['crs']
            _ACTIVE_RASTER=ds
            _ACTIVE_NAME=name
            yield
    finally:
        _ACTIVE_RASTER=None
        _ACTIVE_NAME=None
        if selected_path==local:
            local.unlink(missing_ok=True)
            print(f'RELEASED local SSD for {name}',flush=True)

def make_training_cache(model):
    assert model=='M1'
    selected=[]
    for (site,kind),grp in candidate.groupby(['site','kind']):
        if site==TEST_SITE or kind!='partial':continue
        limit=MAX_PARTIAL_PER_MOSAIC*(3 if site=='MOS' else 1)
        subset=grp.sample(n=min(limit,len(grp)),random_state=SEED)
        selected.extend(subset.itertuples(index=False))
    selected_by_mosaic={}
    for rec in selected:
        selected_by_mosaic.setdefault(rec.mosaic,[]).append(rec)
    print('Selected M1 windows by mosaic:',flush=True)
    display(pd.DataFrame([
        dict(mosaic=name,site=geo[name]['site'],selected=len(rows))
        for name,rows in sorted(selected_by_mosaic.items())]))
    chunk_size=100
    parts_dir=CACHE_DIR/f'partial_{CACHE_KEY[:20]}'
    parts_dir.mkdir(parents=True,exist_ok=True)
    all_chips=[]
    t0=time.perf_counter()
    for mosaic_i,(name,records) in enumerate(sorted(selected_by_mosaic.items())):
        chunks=[]
        missing=0
        for offset in range(0,len(records),chunk_size):
            stop=min(offset+chunk_size,len(records))
            part=parts_dir/f'mosaic{mosaic_i:02d}_{offset:05d}_{stop:05d}.npz'
            marker=parts_dir/f'empty_mosaic{mosaic_i:02d}_{offset:05d}_{stop:05d}.txt'
            old=None
            if part.exists():
                try:
                    old=_unpack_chips(part,CACHE_KEY)
                except Exception as exc:
                    print('Discarding invalid chunk:',part.name,repr(exc),flush=True)
                    part.unlink(missing_ok=True)
            if old is None and marker.exists() and marker.read_text()==CACHE_KEY:
                old=[]
            if old is None:missing+=1
            chunks.append((offset,stop,part,marker,old))
        print(f'M1 {name}: {len(records)} candidates; '
              f'{len(chunks)-missing}/{len(chunks)} cached chunks',flush=True)
        if missing:
            with staged_mosaic(name):
                for offset,stop,part,marker,old in chunks:
                    if old is not None:continue
                    fresh=[]
                    for rec in records[offset:stop]:
                        chip=training_crop(rec,model)
                        if chip is not None:fresh.append(chip)
                    if fresh:
                        _pack_chips(part,fresh,CACHE_KEY)
                    else:
                        marker.write_text(CACHE_KEY)
                    print(f'M1 {name}: {stop}/{len(records)} attempted, '
                          f'{len(fresh)} usable in chunk, '
                          f'{(time.perf_counter()-t0)/60:.1f} min elapsed',flush=True)
        for offset,stop,part,marker,old in chunks:
            if old is not None:all_chips.extend(old)
            elif part.exists():all_chips.extend(_unpack_chips(part,CACHE_KEY))
            else:assert marker.exists() and marker.read_text()==CACHE_KEY
        print(f'M1 completed {name}; total usable chips {len(all_chips)}',flush=True)
    sites=pd.Series([x['site'] for x in all_chips]).value_counts().to_dict()
    print('M1 chips by site:',sites,'| total:',len(all_chips),
          '| elapsed minutes:',round((time.perf_counter()-t0)/60,2),flush=True)
    assert len(all_chips)>15
    assert all(chip['site']!=TEST_SITE for chip in all_chips), 'LOSO leakage in training chips'
    return all_chips

class CropDataset(Dataset):
    def __init__(self,rows):self.rows=rows
    def __len__(self):return len(self.rows)
    def __getitem__(self,i):
        q=self.rows[i];x=q['image'].copy();y=q['target'].copy();w=q['weight'].copy()
        k=int(torch.randint(0,4,(1,)).item())
        x=np.rot90(x,k);y=np.rot90(y,k);w=np.rot90(w,k)
        if torch.rand(1).item()<.5:x=x[:,::-1];y=y[:,::-1];w=w[:,::-1]
        return (torch.from_numpy(np.ascontiguousarray(x.transpose(2,0,1))),
                torch.from_numpy(np.ascontiguousarray(y)),torch.from_numpy(np.ascontiguousarray(w)))


# Cache only native, non-augmented crops; record source-file metadata and training rules.
import hashlib
CACHE_DIR=ROOT/'Outputs/V3/02A_partial_only_LOSO_no_MOS34'/'chip_cache'
CACHE_DIR.mkdir(parents=True,exist_ok=True)
def fingerprint():
    deps=[]
    paths=[REF/'01_candidate_windows.csv',
           ROOT/'Outputs/V3/00_input_audit/00_input_manifest.json',
           A['roi'],A['exhaustive_annotations']['complete_roi_boundaries']]
    paths += [A['shared_annotations'][key] for key in sorted(A['shared_annotations'])]
    paths += [v['path'] for v in mosaics.values()]
    for path in paths:
        path=Path(path)
        more=[path.with_suffix(e) for e in ('.shp','.shx','.dbf','.prj','.cpg')
              if path.with_suffix(e).exists()] if path.suffix=='.shp' else [path]
        for p in more:
            deps.append([str(p),p.stat().st_size,p.stat().st_mtime_ns])
    spec=dict(version='M1_LOSO_noMOS34_current_ref_v2_strict_site_holdout',dependencies=deps,
              seed=SEED,crop=CROP,holdout_ids=sorted(HOLDOUT),
              max_partial=MAX_PARTIAL_PER_MOSAIC,test_site=TEST_SITE,
              conflict_weight_zero=True)
    return hashlib.sha256(json.dumps(spec,sort_keys=True).encode()).hexdigest()
CACHE_KEY=fingerprint()
CACHE_FILE=CACHE_DIR/f'M1_{CACHE_KEY[:20]}.npz'
def get_m1_cache():
    if CACHE_FILE.exists():
        started=time.perf_counter()
        try:
            result=_unpack_chips(CACHE_FILE,CACHE_KEY)
            assert len(result)>15
            assert all(chip['site']!=TEST_SITE for chip in result), 'LOSO leakage in cached chips'
            print('M1 FULL CACHE HIT:',len(result),'chips restored in',
                  round(time.perf_counter()-started,1),'seconds',flush=True)
            return result
        except Exception as e:
            print('Invalid full M1 cache, falling back to partial chunks:',repr(e),flush=True)
            CACHE_FILE.unlink(missing_ok=True)
    print('M1 cache miss: preparing source crops once.',flush=True)
    started=time.perf_counter()
    result=make_training_cache('M1')
    _pack_chips(CACHE_FILE,result,CACHE_KEY)
    print('M1 CACHE SAVED:',len(result),'chips in',
          round(time.perf_counter()-started,1),'seconds',flush=True)
    return result
```

## 05 — Four held-out sites: train M1, then evaluate independent partial tree labels

Each site model is fitted independently; MOS 1–2 and MOS 5–6 are held out together. AUC/AP use **only known positive/negative tree polygons** in the held-out site's candidate windows. Those windows may overlap, so treat scores as patch-sampled pixel discrimination, not an unbiased site-wide census. No exhaustive annotations are used for targets.


```python
import time
from contextlib import nullcontext
USE_AMP=(device=='cuda')
# float16 is fast on both T4 and L4; BCE is evaluated in fp32 for stability.
scaler=torch.amp.GradScaler('cuda',enabled=USE_AMP)
print('CUDA mixed precision:',USE_AMP,'| GPU:',torch.cuda.get_device_name(0),flush=True)

def fit_m1():
    ckpt=OUT/'M1_resnet18_strictLOSO_model.pt'
    torch.manual_seed(SEED)
    model=ResNet18UNet().to(device)
    if ckpt.exists():
        model.load_state_dict(torch.load(ckpt,map_location=device,weights_only=True))
        print('Reusing completed ResNet18 checkpoint:',ckpt,flush=True)
        return model
    print('M1: checking persisted partial-only cache...',flush=True)
    prep=time.perf_counter()
    cache=get_m1_cache()
    print(f'M1: {len(cache)} usable training chips; prep {time.perf_counter()-prep:.1f}s',flush=True)
    assert all(chip['site']!=TEST_SITE for chip in cache), 'LOSO leakage: abort training'
    print('LOSO training sites:', sorted(set(chip['site'] for chip in cache)), '| held out:', TEST_SITE, flush=True)
    loader=DataLoader(CropDataset(cache),batch_size=BATCH,shuffle=True,num_workers=0,
                      pin_memory=True)
    optim=torch.optim.AdamW(model.parameters(),lr=2e-4,weight_decay=1e-4)
    history=[]
    t0=time.perf_counter()
    for epoch in range(EPOCHS):
        model.train();losses=[];t_epoch=time.perf_counter()
        for step,(x,y,w) in enumerate(loader,1):
            x=x.to(device,non_blocking=True)
            y=y.to(device,non_blocking=True)
            w=w.to(device,non_blocking=True)
            optim.zero_grad(set_to_none=True)
            with torch.autocast(device_type='cuda',dtype=torch.float16,enabled=USE_AMP):
                logits=model(x)
            loss=masked_loss(logits.float(),y,w)
            scaler.scale(loss).backward()
            scaler.step(optim)
            scaler.update()
            losses.append(float(loss.detach()))
        elapsed=time.perf_counter()-t_epoch
        avg=float(np.mean(losses))
        history.append(dict(epoch=epoch+1,masked_BCE=avg,seconds=elapsed))
        pd.DataFrame(history).to_csv(OUT/'M1_loss.csv',index=False)
        print(f'M1 epoch {epoch+1:02d}/{EPOCHS}: BCE={avg:.4f}, '
              f'{elapsed:.1f}s, total {(time.perf_counter()-t0)/60:.1f} min',
              flush=True)
    torch.save(model.state_dict(),ckpt)
    print('ResNet18 training complete; checkpoint:',ckpt,flush=True)
    del cache
    return model

print('Training function ready; execution is controlled by the LOSO loop below.',flush=True)
```

```python
from sklearn.metrics import roc_auc_score,average_precision_score
import gc
SITES=['Baego','MOS','Yangambi','Yoko']
BASE_OUT=ROOT/'Outputs/V3/02A_partial_only_LOSO_no_MOS34'
def evaluate_partial_heldout(model,test_site):
    model.eval()
    positives=np.zeros(256,dtype=np.float64)
    counts=np.zeros(256,dtype=np.float64)
    seen=0;masked_pixels=0
    for name in sorted(k for k,v in geo.items() if v['site']==test_site):
        records=candidate[(candidate.mosaic==name)&(candidate.kind=='partial')]
        # SSD staging also speeds up held-out inference.
        with staged_mosaic(name):
            for rec in records.itertuples(index=False):
                if rec.height<CROP or rec.width<CROP:continue
                rr=int(rec.row)+(int(rec.height)-CROP)//2
                cc=int(rec.col)+(int(rec.width)-CROP)//2
                win=Window(cc,rr,CROP,CROP)
                d=geo[name]
                y_pos=burn_window(d['tree_positive'],name,win)
                y_neg=burn_window(d['tree_negative'],name,win)
                label=(y_pos|y_neg)&~(y_pos&y_neg)
                label &= burn_window(d['roi_local'].geometry,name,win)
                rgb,valid=rgb_for_window(name,win)
                label &= valid
                if not label.any():continue
                with torch.inference_mode():
                    x=torch.from_numpy(rgb.transpose(2,0,1).copy())[None].to(device)
                    probs=torch.sigmoid(model(x))[0].detach().cpu().numpy()
                y=y_pos[label].astype('float64')
                p=probs[label]
                bix=np.minimum((p*256).astype('int32'),255)
                counts+=np.bincount(bix,minlength=256)
                positives+=np.bincount(bix,weights=y,minlength=256)
                seen+=1;masked_pixels+=int(label.sum())
        print(test_site,name,'evaluated windows:',seen,'pixels:',masked_pixels,flush=True)
    neg=counts-positives
    tp=float(positives.sum());tn=float(neg.sum())
    assert tp>0 and tn>0,(test_site,tp,tn)
    auc=float(np.dot(positives,np.cumsum(neg)-.5*neg)/(tp*tn))
    cumpos=np.cumsum(positives[::-1]);cumneg=np.cumsum(neg[::-1])
    ap=float(np.sum(positives[::-1]/tp*cumpos/np.maximum(cumpos+cumneg,1)))
    return dict(site=test_site,auc=auc,ap=ap,evaluated_windows=seen,
                known_pixels=int(masked_pixels),positive_pct=100*tp/(tp+tn))

METRICS=[]
RUN_SITES=SITES  # all four independent site holdouts
for TEST_SITE in RUN_SITES:
    OUT=BASE_OUT/TEST_SITE
    OUT.mkdir(parents=True,exist_ok=True)
    CACHE_KEY=fingerprint()
    CACHE_FILE=CACHE_DIR/f'M1_{CACHE_KEY[:20]}.npz'
    print('=== LOSO partial-only: TEST SITE',TEST_SITE,'===',flush=True)
    model1=fit_m1()
    metrics=evaluate_partial_heldout(model1,TEST_SITE)
    METRICS.append(metrics)
    pd.DataFrame(METRICS).to_csv(BASE_OUT/'02A_LOSO_partial_scores.csv',index=False)
    del model1;torch.cuda.empty_cache();gc.collect()
display(pd.DataFrame(METRICS).round(4))
print(f'Saved/verified {len(METRICS)} LOSO M1 checkpoints under',BASE_OUT)
```

```python
from sklearn.metrics import roc_auc_score,average_precision_score,r2_score,mean_absolute_error
# Combine 256-bin histograms of *all* valid image/reference pixels rather than
# averaging per-circle sampled AUCs. This evaluates pixel discrimination; the
# histograms provide a ~1/256 probability-resolution approximation to exact AUC.
def binned_metrics(ids):
    cnt=np.sum([hist1[str(int(cid))+'_count'] for cid in ids],axis=0).astype('float64')
    pos=np.sum([hist1[str(int(cid))+'_positive'] for cid in ids],axis=0).astype('float64')
    neg=cnt-pos
    assert (neg>=0).all()
    npos=pos.sum();nneg=neg.sum()
    if npos==0 or nneg==0:return dict(auc=np.nan,ap=np.nan,pixels=int(cnt.sum()))
    # Tie-aware Mann-Whitney: negatives with lower scores + half tied negatives.
    auc=float(np.dot(pos,np.cumsum(neg)-.5*neg)/(npos*nneg))
    # AP uses descending score order and precision at each positive score.
    pr=np.cumsum(pos[::-1]);fp=np.cumsum(neg[::-1])
    ap=float(np.sum((pos[::-1]/npos)*(pr/np.maximum(pr+fp,1))))
    return dict(auc=auc,ap=ap,pixels=int(cnt.sum()),positive_pct=100*npos/(npos+nneg))
report=[]
for site,part in m1.groupby('site'):
    report.append(dict(group_type='site',group=site,circles=len(part),
                       **binned_metrics(part.circle_id),
                       MAE_fraction_pp=float(part.bias_pp.abs().mean())))
for mosaic,part in m1.merge(links[['circle_id','mosaic']],on='circle_id').groupby('mosaic'):
    report.append(dict(group_type='mosaic',group=mosaic,circles=len(part),
                       **binned_metrics(part.circle_id),
                       MAE_fraction_pp=float(part.bias_pp.abs().mean())))
report.append(dict(group_type='overall',group='ALL 30 independent circles',
                   circles=len(m1),**binned_metrics(m1.circle_id),
                   MAE_fraction_pp=float(m1.bias_pp.abs().mean())))
REPORT=pd.DataFrame(report).sort_values(['group_type','group'])
assert m1.circle_id.nunique()==30
assert set(m1.circle_id.astype(int))==set(links.circle_id.astype(int))
print('=== INDEPENDENT M1 PIXEL DISCRIMINATION: 30 exhaustive circles ===')
display(REPORT.round(4))
print('=== CIRCLE-LEVEL FRACTION: ALL 30 (independent) ===')
print('R²',round(r2_score(m1.observed_fraction,m1.predicted_fraction),4),
      '| MAE pp',round(mean_absolute_error(m1.observed_fraction,m1.predicted_fraction),3))
REPORT.to_csv(REPORT_OUT/'02A_M1_roc_auc_by_site_mosaic.csv',index=False)
m1.to_csv(REPORT_OUT/'02A_M1_all30_circle_predictions.csv',index=False)

from sklearn.metrics import mean_squared_error
summary_fraction=pd.DataFrame([dict(group='ALL 30 circles',n_circles=len(m1),
    R2=r2_score(m1.observed_fraction,m1.predicted_fraction),
    MAE_pp=mean_absolute_error(m1.observed_fraction,m1.predicted_fraction),
    RMSE_pp=np.sqrt(mean_squared_error(m1.observed_fraction,m1.predicted_fraction)),
    bias_pp=float(m1.bias_pp.mean()))])
display(summary_fraction.round(4))
summary_fraction.to_csv(REPORT_OUT/'02A_M1_fraction_summary.csv',index=False)
assert set(m1.circle_id.astype(int))==set(links.circle_id.astype(int))
print('NB02A complete: 4 M1 site holdouts; 30 independent exhaustive circles')
```

## 08 — Spatial occurrence diagnostics (M1, independent 30-circle holdout)

**Evaluation only; no retraining.** Reuse the four saved ResNet18 LOSO checkpoints and the NB01 exhaustive reference. Report per-circle native-pixel IoU, Dice/F1, precision and recall at the predeclared **0.50** probability cutoff; evaluate Musanga occurrence on **5, 10 and 20 m** cells using a clearly marked diagnostic **10% fractional-cover cutoff** and at least 50% valid coverage of each cell. AUC/AP and fractional-cover metrics above remain unchanged.

Save all-circle 5 m spatial maps and a representative native-resolution window comparison (RGB, reference, probability, FP/FN) per circle. Outputs are written to a separate `02A_spatial_validation` folder. Grid cells use each circle's projected bounding-box origin. These are diagnostic occurrence maps, not object/crown-matching scores.

```python
# Block 08 — independent spatial diagnostics; uses already-trained M1 checkpoints
from pathlib import Path
from rasterio.windows import Window
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import torch, gc
from IPython.display import display

SPATIAL_OUT=ROOT/'Outputs/V3/02A_partial_only_LOSO_no_MOS34'/'02A_spatial_validation'
SPATIAL_OUT.mkdir(parents=True,exist_ok=True)
SPATIAL_PIXEL_THRESHOLD=0.50
SPATIAL_GRID_SCALES=(5,10,20)       # metres, CRS must be metric
SPATIAL_OCCURRENCE_MIN_FRACTION=0.10  # diagnostic, fixed a priori
SPATIAL_MIN_VALID_CELL_FRAC=0.50
# Set to False to avoid generating the 30 PNG plots; metrics still computed.
SPATIAL_SAVE_PLOTS=True
assert all(geo[k]['crs'].is_projected for k in geo)
assert set(VAL_WINDOWS.circle_id.dropna().astype(int))==set(links.circle_id.astype(int))
assert len(links)==30

def _spatial_confusion(yhat,truth):
    tp=int(np.count_nonzero(yhat & truth))
    fp=int(np.count_nonzero(yhat & ~truth))
    fn=int(np.count_nonzero(~yhat & truth))
    tn=int(np.count_nonzero(~yhat & ~truth))
    return tp,fp,fn,tn

def _spatial_scores(tp,fp,fn,tn,prefix=''):
    return {prefix+'precision':tp/(tp+fp) if tp+fp else np.nan,
            prefix+'recall':tp/(tp+fn) if tp+fn else np.nan,
            prefix+'iou':tp/(tp+fp+fn) if tp+fp+fn else np.nan,
            prefix+'dice_f1':2*tp/(2*tp+fp+fn) if 2*tp+fp+fn else np.nan,
            prefix+'tp':tp,prefix+'fp':fp,prefix+'fn':fn,prefix+'tn':tn}

def _spatial_grid_setup(bound,scale):
    xmin,ymin,xmax,ymax=bound
    nx=max(1,int(np.ceil((xmax-xmin)/scale)))
    ny=max(1,int(np.ceil((ymax-ymin)/scale)))
    n=nx*ny
    return dict(xmin=xmin,ymin=ymin,nx=nx,ny=ny,scale=scale,
                valid=np.zeros(n,dtype=np.int64),
                positive=np.zeros(n,dtype=np.int64),
                pred_binary=np.zeros(n,dtype=np.int64),
                pred_sum=np.zeros(n,dtype=np.float64))

def _spatial_add(g,xx,yy,truth,pred):
    ix=np.floor((xx-g['xmin'])/g['scale']).astype(np.int32)
    iy=np.floor((yy-g['ymin'])/g['scale']).astype(np.int32)
    inside=(ix>=0)&(ix<g['nx'])&(iy>=0)&(iy<g['ny'])
    if not inside.any():return
    idx=iy[inside].astype(np.int64)*g['nx']+ix[inside]
    n=g['nx']*g['ny']
    g['valid']+=np.bincount(idx,minlength=n)
    g['positive']+=np.bincount(idx,weights=truth[inside],minlength=n).astype('int64')
    g['pred_binary']+=np.bincount(idx,weights=(pred[inside]>=SPATIAL_PIXEL_THRESHOLD),minlength=n).astype('int64')
    g['pred_sum']+=np.bincount(idx,weights=pred[inside],minlength=n)

def _spatial_grid_result(g,pixel_area):
    good=g['valid']*pixel_area >= SPATIAL_MIN_VALID_CELL_FRAC*g['scale']**2
    valid=np.maximum(g['valid'],1)
    obs=g['positive']/valid
    pred=g['pred_sum']/valid
    # Compare fractional cover within valid cells; the 10% presence threshold is diagnostic.
    gt=obs[good]>=SPATIAL_OCCURRENCE_MIN_FRACTION
    pp=pred[good]>=SPATIAL_OCCURRENCE_MIN_FRACTION
    tp,fp,fn,tn=_spatial_confusion(pp,gt)
    return dict(good=good,observed=obs,predicted=pred,
                summary=dict(scale_m=g['scale'],n_valid_cells=int(good.sum()),
                             **_spatial_scores(tp,fp,fn,tn)))

def _spatial_plot(cid,site,name,g,preview):
    z=_spatial_grid_result(g,geo[name]['pixel_area'])
    ny,nx=g['ny'],g['nx']
    mask=z['good'].reshape(ny,nx)
    obs=np.where(mask,z['observed'].reshape(ny,nx),np.nan)
    pred=np.where(mask,z['predicted'].reshape(ny,nx),np.nan)
    diff=pred-obs
    fig,axes=plt.subplots(2,4,figsize=(17,8),constrained_layout=True)
    ims=[obs,pred,diff,np.where(mask,(pred>=SPATIAL_OCCURRENCE_MIN_FRACTION).astype(float)-(obs>=SPATIAL_OCCURRENCE_MIN_FRACTION).astype(float),np.nan)]
    specs=[(obs,'Observed 5 m fraction',0,1,'viridis'),
           (pred,'Predicted mean probability',0,1,'viridis'),
           (diff,'Predicted – observed',-1,1,'coolwarm'),
           (ims[3],'Occurrence error (-1 missed, +1 false)',-1,1,'bwr')]
    for ax,(v,label,vmin,vmax,cmap) in zip(axes[0],specs):
        im=ax.imshow(v,origin='lower',cmap=cmap,vmin=vmin,vmax=vmax,interpolation='nearest')
        ax.set_title(label);ax.set_axis_off()
        fig.colorbar(im,ax=ax,fraction=0.046,pad=0.04)
    rgb,ref,prob,valid=preview
    false=np.ma.masked_all(ref.shape,dtype='float32')
    false[valid]=0
    false[valid & (prob>=SPATIAL_PIXEL_THRESHOLD) & ~ref]=1
    false[valid & (prob<SPATIAL_PIXEL_THRESHOLD) & ref]=-1
    four=[(rgb,'RGB: representative window',None,None,None),
          (np.where(valid,ref.astype(float),np.nan),'Exhaustive reference',0,1,'gray'),
          (np.where(valid,prob,np.nan),'ResNet18 probability',0,1,'viridis'),
          (false,'FP red / FN blue',-1,1,'bwr')]
    for ax,(v,label,vmin,vmax,cmap) in zip(axes[1],four):
        ax.imshow(v,cmap=cmap,vmin=vmin,vmax=vmax)
        ax.set_title(label);ax.set_axis_off()
    fig.suptitle(f'Circle {cid} · {site} · {name} · independent M1',fontsize=14)
    fig.savefig(SPATIAL_OUT/f'circle_{cid:02d}_spatial.png',dpi=135)
    plt.close(fig)

SPATIAL_ROWS=[]
SPATIAL_GRID_ROWS=[]
for site in ['Baego','MOS','Yangambi','Yoko']:
    ckpt=ROOT/'Outputs/V3/02A_partial_only_LOSO_no_MOS34'/site/'M1_resnet18_strictLOSO_model.pt'
    assert ckpt.exists(),ckpt
    model=ResNet18UNet().to(device).eval()
    model.load_state_dict(torch.load(ckpt,map_location=device,weights_only=True))
    for row in links.loc[links.site==site].itertuples(index=False):
        cid=int(row.circle_id);name=row.mosaic;d=geo[name]
        circle=circles.loc[circles.original_row==cid].to_crs(d['crs'])
        bounds=circle.total_bounds
        grids={scale:_spatial_grid_setup(bounds,scale) for scale in SPATIAL_GRID_SCALES}
        n_pixel=tp=fp=fn=tn=0
        preview=None
        windows=VAL_WINDOWS[(VAL_WINDOWS.circle_id==cid)&(VAL_WINDOWS.mosaic==name)]
        assert len(windows)>0,(site,cid,name)
        with staged_mosaic(name):
            for rec in windows.itertuples(index=False):
                win=Window(int(rec.col),int(rec.row),int(rec.width),int(rec.height))
                valid=burn_window(circle.geometry,name,win)
                valid &= burn_window(d['roi_local'].geometry,name,win)
                rgb,rgb_valid=rgb_for_window(name,win)
                valid &= rgb_valid
                if not valid.any():continue
                reference=burn_window(d['positive_local'].geometry,name,win)
                with torch.inference_mode():
                    x=torch.from_numpy(np.ascontiguousarray(rgb.transpose(2,0,1)))[None].to(device)
                    probability=torch.sigmoid(model(x))[0].float().cpu().numpy()
                tr=reference[valid];pr=probability[valid]
                a,b,c,e=_spatial_confusion(pr>=SPATIAL_PIXEL_THRESHOLD,tr)
                tp+=a;fp+=b;fn+=c;tn+=e;n_pixel+=len(tr)
                if preview is None:
                    preview=(rgb.copy(),reference.copy(),probability.copy(),valid.copy())
                rr,cc=np.nonzero(valid)
                transf=rasterio.windows.transform(win,d['transform'])
                xx=transf.c+(cc+.5)*transf.a+(rr+.5)*transf.b
                yy=transf.f+(cc+.5)*transf.d+(rr+.5)*transf.e
                for grid in grids.values():
                    _spatial_add(grid,xx,yy,tr,pr)
        assert n_pixel>0,(site,cid)
        SPATIAL_ROWS.append(dict(site=site,mosaic=name,circle_id=cid,valid_pixels=n_pixel,
                                 **_spatial_scores(tp,fp,fn,tn)))
        for scale,g in grids.items():
            result=_spatial_grid_result(g,abs(d['transform'].a*d['transform'].e-d['transform'].b*d['transform'].d))
            SPATIAL_GRID_ROWS.append(dict(site=site,mosaic=name,circle_id=cid,**result['summary']))
        if SPATIAL_SAVE_PLOTS and preview is not None:
            # Use 5-m grids for full-circle maps; native resolution for representative window
            g=grids[5]
            d['pixel_area']=abs(d['transform'].a*d['transform'].e-d['transform'].b*d['transform'].d)
            _spatial_plot(cid,site,name,g,preview)
        pd.DataFrame(SPATIAL_ROWS).to_csv(SPATIAL_OUT/'02A_M1_circle_pixel_overlap.csv',index=False)
        pd.DataFrame(SPATIAL_GRID_ROWS).to_csv(SPATIAL_OUT/'02A_M1_circle_grid_occurrence.csv',index=False)
        print(f'SPATIAL M1 {site} circle {cid}: pixel IoU={SPATIAL_ROWS[-1]["iou"]:.3f}, '
              f'F1={SPATIAL_ROWS[-1]["dice_f1"]:.3f}; '
              f'{len(windows)} windows',flush=True)
    del model
    torch.cuda.empty_cache()
    gc.collect()
pixel_df=pd.DataFrame(SPATIAL_ROWS)
grid_df=pd.DataFrame(SPATIAL_GRID_ROWS)
assert len(pixel_df)==30 and pixel_df.circle_id.is_unique
assert len(grid_df)==90
summary=[]
for site,frame in [('ALL',pixel_df),*list(pixel_df.groupby('site'))]:
    summary.append(dict(site=site,circles=len(frame),
                        **_spatial_scores(*(int(frame[k].sum()) for k in ['tp','fp','fn','tn']))))
summary_df=pd.DataFrame(summary)
display(summary_df.round(4))
display(grid_df.groupby(['scale_m']).agg(circles=('circle_id','nunique'),
                 mean_IoU=('iou','mean'),mean_F1=('dice_f1','mean'),
                 mean_precision=('precision','mean'),mean_recall=('recall','mean')).round(4))
summary_df.to_csv(SPATIAL_OUT/'02A_M1_pixel_overlap_summary.csv',index=False)
print('Spatial validation complete:',SPATIAL_OUT)
```
